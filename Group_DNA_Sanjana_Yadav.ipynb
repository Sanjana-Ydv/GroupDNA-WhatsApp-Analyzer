```python
with open('hostel_bois.txt', 'r', encoding='utf-8') as f:
  lines = f.readlines()
  print(f"Total lines in file: {len(lines)}")
```

```python
messages = []
system_count = 0
media_count = 0
deleted_count = 0

for line in lines:
    line = line.strip()
    if not line:
        continue

    parts = line.split(' - ', 1)
    if len(parts) != 2:
        system_count += 1
        continue

    timestamp_str = parts[0]
    rest = parts[1]

    sender_text_parts = rest.split(': ', 1)
    if len(sender_text_parts) != 2:
        system_count += 1
        continue

    sender = sender_text_parts[0].strip()
    text = sender_text_parts[1].strip()

    if text == "<Media Omitted>":
        media_count += 1
    elif text == "This message was deleted":
        deleted_count += 1
    else:
        messages.append({
                'timestamp': timestamp_str,
                'sender': sender,
                'text': text
        })

print(f"Total parsed messages: {len(messages)}")
print(f"Total system messages: {system_count}")
print(f"Total media messages: {media_count}")
print(f"Total deleted messages: {deleted_count}")
```

```python
from datetime import datetime
person_counts={}

for msg in messages:
  sender= msg['sender']
  if sender in person_counts:
    person_counts[sender] += 1
  else:
    person_counts[sender] = 1


total_messages= len(messages)
participants= list(person_counts.keys())
print(f"Total participants: {len(participants)}")
print(f"Total messages: {total_messages}")
def get_counts(pair):
  return pair [1]
sorted_people= sorted(person_counts.items(), key=get_counts, reverse=True)

print("Messages per person")
for name, count in sorted_people:
  percent= (count/ total_messages)* 100
  print(f"{name:<10} {count:<5} ({percent:5.1f}%)")






```

```python

day_counts = {}
hour_counts = {}

for msg in messages:

    date_parts, time_part = msg['timestamp'].split(', ')
    hour = time_part.split(':')[0]

    day_counts[date_parts] = day_counts.get(date_parts, 0) + 1
    hour_counts[hour] = hour_counts.get(hour, 0) + 1

busiest_day = max(day_counts, key=day_counts.get)
busiest_hour = max(hour_counts, key=hour_counts.get)

print(f"Busiest day: {busiest_day} ({day_counts[busiest_day]} messages)")
print(f"Busiest hour: {busiest_hour}:00 ({hour_counts[busiest_hour]} messages)")
```

```python
import numpy as np
participant_list = sorted(set(msg['sender'] for msg in messages))

heatmap = np.zeros((len(participant_list), 24))

for msg in messages:
    sender = msg['sender']
    time_part = msg['timestamp'].split(", ") [1]
    hour = int(time_part.split(":")[0])

    person_index = participant_list.index(sender)
    heatmap[person_index, hour] += 1

print(heatmap)
```

```python
def shade(value, row_max):
    if row_max == 0:
        return ". "
    ratio = value / row_max
    if ratio <= 0.25:
        return ".  "
    elif ratio <= 0.5:
        return "\u2591  "
    elif ratio <= 0.75:
        return "\u2592"
    else:
        return "\u2588"
print("ACTIVITY HEATMAP (by hour)")
print("     "+ "  ".join(f"{h:02d}:00" for h in range(2, 24, 2)))

for name in participant_list:
    row_max = heatmap[participant_list.index(name)].max()
    line = f"{name:<6}"
    for hour in range(2, 24, 2):
        line += shade(value=heatmap[participant_list.index(name)][hour], row_max=row_max)
    print(line)
```

```python
stop_words = {"is", "i", "the", "a", "or", "and", "to", "of", "in", "on", "for", "am", "are", "you", "it", "this", "was", "how", "guys", "so", "about", "today", "at", "my", "he", "his", "her", "she", "we", "us"}

words_counts ={}

for msg in messages:
   text = msg["text"]
   if text == "<Media Omitted>" or text == "This message was deleted":
    continue
   word_list = text.lower().split()
   for word in word_list:

    clean_word = word.strip(".,?!:;()-")
    if clean_word == "" or clean_word in stop_words:
        continue
    words_counts[clean_word] = words_counts.get(clean_word, 0) + 1

def get_count(pair):
  return pair[1]

sorted_words = sorted(words_counts.items(), key=get_count, reverse=True)
top_10 = sorted_words[:10]

if top_10:
     max_count = top_10[0][1]

     print("THIS GROUP'S FAV WORDS")
     for word, count in top_10:
        bar_length = int((count / max_count) * 20)
        bar = " " * bar_length
        print(f"{word:<10} {bar:<20} {count}")
else:
    print("No Words To Analyze.")
```

```python
from datetime import datetime, timedelta

response_gaps = {}

for i in range (1, len(messages)):
  prev_msg = messages[i-1]
  curr_msg = messages[i]

  if curr_msg['sender'] == prev_msg['sender']:
    continue

  t1 = datetime.strptime(prev_msg["timestamp"], '%d/%m/%y, %H:%M')
  t2 = datetime.strptime(curr_msg["timestamp"], '%d/%m/%y, %H:%M')
  gap_seconds = (t2 - t1).total_seconds()

  sender = curr_msg['sender']
  if sender not in response_gaps:
      response_gaps[sender] = []
  response_gaps[sender].append(gap_seconds)

avg_response ={}
for person, gaps in response_gaps.items():
    avg_response[person] = sum(gaps) / len(gaps)

def get_avg(pair):
    return pair[1]

sorted_by_speed = sorted(avg_response.items(), key=get_avg)

fastest = sorted_by_speed[0]
slowest = sorted_by_speed[-1]

print("PATTERN OF RESPONSE")
print(f"Fastest replier: {fastest[0]} ({fastest[1]/60:.1f} minutes)")
print(f"Slowest replier: {slowest[0]} ({slowest[1]/3600:.1f} hours)")

active_dates =  {person:set() for person in participant_list}
all_dates = []

for msg in messages:
  timestamp_dt = datetime.strptime(msg['timestamp'], '%d/%m/%y, %H:%M')
  msg_date = timestamp_dt.date()
  active_dates[msg['sender']].add(msg_date)
  all_dates.append(msg_date)
if all_dates:
    start_date = min(all_dates)
    end_date = max(all_dates)
    total_days = (end_date - start_date).days + 1
else:
    start_date = None
    end_date = None
    total_days = 0

print("LONGEST SILENCE ZZZ")
if total_days > 0:
  for person in participant_list:
    their_dates = active_dates.get(person, set())
    longest_streak = 0
    current_streak = 0
    for day_num in range(total_days):
        this_day = start_date + timedelta(days=day_num)
        if this_day in their_dates:
            current_streak = 0
        else:
            current_streak +=1
            longest_streak = max(longest_streak, current_streak)
    print(f"{person:<10} {longest_streak} days")
else:
   print("No messages to analyze for longest silence.")
```

```python
def chat_machine_score(person, all_messages):
  burst_lengths = []
  current_burst = 0
  for msg in all_messages:
        if msg['sender'] == person:
            current_burst += 1
        else:
           if current_burst > 0:
              burst_lengths.append(current_burst)
              current_burst = 0
  if current_burst > 0:
       burst_lengths.append(current_burst)
  if len(burst_lengths) == 0:
       return 0
  return sum(burst_lengths) / len(burst_lengths)

def gaurdian_angel_score(person_messages):
    caring_words = ["okay", "eat", "sleep", "take care", "safe", "are you", "reminder", "please", "drink water", "don't forget"]
    if len(person_messages) == 0:
        return 0
    score = 0
    for msg in person_messages:
      text = msg['text'].lower()
      for word in caring_words:
           if word in text:
            score += 1
    return(score / len(person_messages)) * 100

def insomniac_score(person_messages):
  if len(person_messages) == 0:
    return 0
  night_count = 0
  for msg in person_messages:
      hour = int(msg["timestamp"].split(", ")[1].split(":")[0])
      if hour >=23 or hour <=4:
         night_count += 1
  return(night_count / len(person_messages)) * 100

def narrator_score(person_messages):
  if len(person_messages) == 0:
      return 0
  total_words = 0
  for msg in person_messages:
    total_words += len(msg["text"].split())
  return total_words / len(person_messages)

def dramebaaz_score(person_messages):
   if len(person_messages) == 0:
       return 0
   drama_count = 0
   for msg in person_messages:
      text = msg["text"]
      letters_only = "".join(c for c in text if c.isalpha())
      is_allcaps = len(letters_only) >=3 and letters_only.isupper()
      has_multi_exclaim = text.count("!") >= 2
      if is_allcaps or has_multi_exclaim:
          drama_count += 1
   return (drama_count / len(person_messages)) * 100

def observer_score(person, active_dates_dict, total_days):
    their_dates = active_dates_dict.get(person, set())
    silent_days = total_days - len(their_dates)
    return (silent_days / total_days) * 100

def joker_score(person_messages):
  if len(person_messages) == 0:
      return 0
  funny_words = ["lol", "lmao", "haha", "rofl", "lmfao"]
  count = 0
  for msg in person_messages:
      text = msg["text"].lower()
      for word in funny_words:
          if word in text:
              count += 1
  return (count / len(person_messages)) * 100


def investigator_score(person_messages):
  if len(person_messages) == 0:
     return 0
  q_count = sum(1 for msg in person_messages if msg["text"].strip().endswith("?"))
  return (q_count / len(person_messages)) * 100

def foodie_score(person_messages):
    food_words = ["chai", "khana", "canteen", "mess", "tapri", "nashta"]
    if len (person_messages) == 0:
        return 0
    count = 0
    for msg in person_messages:
        text = msg["text"].lower()
        for word in food_words:
            if word in text:
                count += 1
    return (count / len(person_messages)) * 100
```

```python
messages_by_person = {}
for msg in messages:
    sender = msg['sender']
    if sender not in messages_by_person:
        messages_by_person[sender] = []
    messages_by_person[sender].append(msg)

archetype_names =["CHAT MACHINE", "GAURDIAN ANGEL", "INSOMNIAC", "NARRATOR", "DRAMEBAAZ", "OBSERVER", "JOKER", "INVESTIGATOR", "FOODIE"]

print("PERSONALITY ARCHETYPES")
for person in participant_list:
    person_messages = messages_by_person[person]
    scores = {
        "CHAT MACHINE": chat_machine_score(person, messages),
        "GAURDIAN ANGEL": gaurdian_angel_score(person_messages),
        "INSOMNIAC": insomniac_score(person_messages),
        "NARRATOR": narrator_score(person_messages),
        "DRAMEBAAZ": dramebaaz_score(person_messages),
        "OBSERVER": observer_score(person, active_dates, total_days),
        "JOKER": joker_score(person_messages),
        "INVESTIGATOR": investigator_score(person_messages),
        "FOODIE": foodie_score(person_messages)
     }

    thresholds = {
    "CHAT MACHINE": 3, "GAURDIAN ANGEL": 5, "INSOMNIAC":60, "NARRATOR": 30, "DRAMEBAAZ": 30, "OBSERVER": 60, "JOKER": 15, "INVESTIGATOR": 25, "FOODIE": 15
    }

    best_archetype = None
    best_margin = -999
    for name in archetype_names:
        margin = scores[name] - thresholds[name]
        if margin > best_margin:
           best_margin = margin
           best_archetype = name

    print(f"{person:<10} -> THE {best_archetype}  (score: {scores[best_archetype]:.1f})")
```
